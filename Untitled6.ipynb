```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df={
    "color":["red","green","pink","black"]
}
df1=pd.DataFrame(df)
print(df1)
encoder=OneHotEncoder(sparse_output=False)
encoder1=encoder.fit_transform(df1[["color"]])
encoder2=pd.DataFrame(encoder1,columns=encoder.get_feature_names_out(["color"]))


print(encoder2)
```

```python
#LabelEncoder sorts alphabetically
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Create DataFrame
df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Small", "Large"]
})

print("Original Data:")
print(df)

# Create LabelEncoder object
le = LabelEncoder()

# Apply Label Encoding
df["Size"] = le.fit_transform(df["Size"])

print("\nAfter Label Encoding:")
print(df)
```

Dummy Encoding (pd.get_dummies()) is used for nominal categorical data, where there is no natural order, such as Color, City, Country, Department, and Brand. It converts each category into separate 0/1 binary columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "Finance", "HR"]
})

dummy = pd.get_dummies(df["Department"],dtype=int)

print(dummy)
```
