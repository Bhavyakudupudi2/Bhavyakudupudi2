```python
import pandas as pd

data = {
    "Order ID": [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "Date": ["01-09-2026","03-09-2026","05-09-2026","15-09-2026","04-09-2026",
             "02-09-2026","03-09-2026","04-09-2026","02-09-2026","03-09-2026"],
    "Product": ["Laptop","Monitor","Printer","Office Chair","Desk","Chair","Table",
                "Headphones","Keyboard","Mouse"],
    "Category": ["Electronics","Electronics","Electronics","Furniture","Furniture",
                 "Furniture","Furniture","Electronics","Electronics","Electronics"],
    "Quantity": [2,2,1,3,1,2,1,4,3,5],
    "Price": [50000,12000,18000,6000,4300,3600,2900,2200,1500,800],
    "Region": ["Hyderabad","Amalapuram","Vijayawada","Amalapuram","Visakhapatnam",
               "Vijayawada","Visakhapatnam","Vijayawada","Visakhapatnam","Vijayawada"],
    "Sales": [100000,24000,18000,18000,4300,7200,2900,8800,4500,4000]
}

df = pd.DataFrame(data)

print(df)
```

```python
total_Quantity=df["Quantity"].sum()
print("total Quantity sold",total_Quantity)
```

```python
average_sales=df['Sales'].mean()
print(average_sales)
```

```python
max_sales=df['Sales'].max()
print(max_sales)
```

```python
min_sales=df['Sales'].min()
print(min_sales)
```

```python
highest_sales=df.loc[df['Sales'].idxmax()]
print(highest_sales)
```

```python
lowest_sales=df.loc[df['Sales'].idxmin()]
print(lowest_sales)
```

```python
region_sales=df.groupby("Region")['Sales'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
product_sales=df.groupby("Product", as_index=False)["Sales"].sum()
product_sales=product_sales.sort_values("Sales",ascending=False)
print(product_sales)
```

```python
category_sales = df.groupby("Category", as_index=False)["Sales"].sum()
category_sales = category_sales.sort_values("Sales", ascending=False)

print(category_sales)
```

```python
quantity_by_product = df.groupby("Product", as_index=False)["Quantity"].sum()
quantity_by_product = quantity_by_product.sort_values("Quantity", ascending=False)

print(quantity_by_product)
```

```python
print("Highest Sales Region:", df.groupby("Region")["Sales"].sum().idxmax())
print("Highest Sales Product:", df.groupby("Product")["Sales"].sum().idxmax())
print("Highest Sales Category:", df.groupby("Category")["Sales"].sum().idxmax())
```
