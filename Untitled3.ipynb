```python
#We use random_state to get the same train-test split every time we run the code.
import numpy as np
data1=np.arange(20).reshape(2,10).T
data2=np.arange(10)
x_train,x_test,y_train,y_test=train_test_split(data1,data2,random_state=0)
print(x_train)
print()
print(x_test)
print()
print(y_train)
print()
print(y_test)
```

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
d = {
    "area":[1,2,3,4],
    "price":[10,20,30,40]
}
df=pd.DataFrame(d)
x = df.area.values
y = df.price.values
slope,intercept,r_value,p_value,std_err=stats.linregress(x,y)

print(r_value)

# So we can confirm that there is strong a relationship.
```

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
d = {
"area": [1, 2, 3, 4],
"rate": [10, 20, 30, 40]
}
df = pd.DataFrame(d)
X = df.area.values
y = df.rate.values
plt.xlabel("X")
plt.ylabel("y")
plt.scatter(X, y)
plt.show()

#scipy is a Python library used for scientific and mathematical computations.
#stats is a module inside the SciPy library.'''
```
